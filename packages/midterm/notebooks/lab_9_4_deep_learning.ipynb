## 2. MNIST Handwritten Digit Recognition (Fully Connected)

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torchvision import datasets, transforms
from torch.utils.data import DataLoader
import matplotlib.pyplot as plt

# Data transforms
transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.1307,), (0.3081,)),
])

# Download and load MNIST
train_dataset = datasets.MNIST(root="../data", train=True, download=True, transform=transform)
test_dataset = datasets.MNIST(root="../data", train=False, download=True, transform=transform)

train_loader = DataLoader(train_dataset, batch_size=64, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=1000, shuffle=False)

print(f"Training samples: {len(train_dataset)}")
print(f"Test samples: {len(test_dataset)}")
```

```python
# Visualize samples
fig, axes = plt.subplots(2, 5, figsize=(10, 4))
for i, ax in enumerate(axes.flat):
    img, label = train_dataset[i]
    ax.imshow(img.squeeze(), cmap="gray")
    ax.set_title(f"Label: {label}")
    ax.axis("off")
plt.suptitle("MNIST Sample Images")
plt.tight_layout()
plt.savefig("../output/9_4_1_mnist_samples.png", dpi=100, bbox_inches="tight")
plt.show()
```

```python
# Define fully connected network
class FCNet(nn.Module):
    def __init__(self):
        super().__init__()
        self.flatten = nn.Flatten()
        self.fc1 = nn.Linear(784, 256)
        self.fc2 = nn.Linear(256, 128)
        self.fc3 = nn.Linear(128, 10)
        self.relu = nn.ReLU()

    def forward(self, x):
        x = self.flatten(x)
        x = self.relu(self.fc1(x))
        x = self.relu(self.fc2(x))
        return self.fc3(x)


device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
model = FCNet().to(device)
print(f"Device: {device}")
print(f"Parameters: {sum(p.numel() for p in model.parameters()):,}")
```

```python
# Training
criterion = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.001)

train_losses = []
for epoch in range(5):
    model.train()
    epoch_loss = 0
    for batch_idx, (data, target) in enumerate(train_loader):
        data, target = data.to(device), target.to(device)
        optimizer.zero_grad()
        output = model(data)
        loss = criterion(output, target)
        loss.backward()
        optimizer.step()
        epoch_loss += loss.item()

    avg_loss = epoch_loss / len(train_loader)
    train_losses.append(avg_loss)
    print(f"Epoch {epoch + 1}/5 | Loss: {avg_loss:.4f}")
```

```python
# Evaluation
model.eval()
correct, total = 0, 0
with torch.no_grad():
    for data, target in test_loader:
        data, target = data.to(device), target.to(device)
        output = model(data)
        pred = output.argmax(dim=1)
        correct += (pred == target).sum().item()
        total += target.size(0)

accuracy = correct / total
print(f"\nTest Accuracy: {accuracy:.4f} ({correct}/{total})")
```

```python
# Plot training loss
plt.figure(figsize=(7, 4))
plt.plot(range(1, 6), train_losses, marker="o")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title(f"MNIST FC Net Training (Final Acc: {accuracy:.2%})")
plt.grid(True, alpha=0.3)
plt.savefig("../output/9_4_2_mnist_loss.png", dpi=100, bbox_inches="tight")
plt.show()
```

## 3. CNN Image Classification (CIFAR-10)

```python
# Load CIFAR-10
cifar_transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize((0.4914, 0.4822, 0.4465), (0.2470, 0.2435, 0.2616)),
])

cifar_train = datasets.CIFAR10(root="../data", train=True, download=True, transform=cifar_transform)
cifar_test = datasets.CIFAR10(root="../data", train=False, download=True, transform=cifar_transform)

cifar_train_loader = DataLoader(cifar_train, batch_size=64, shuffle=True)
cifar_test_loader = DataLoader(cifar_test, batch_size=256, shuffle=False)

classes = cifar_train.classes
print(f"Classes: {classes}")
print(f"Train: {len(cifar_train)}, Test: {len(cifar_test)}")
```

```python
# Show sample images
fig, axes = plt.subplots(2, 5, figsize=(10, 4))
for i, ax in enumerate(axes.flat):
    img, label = cifar_train[i]
    img_np = img.permute(1, 2, 0).numpy()
    img_np = img_np * np.array([0.2470, 0.2435, 0.2616]) + np.array([0.4914, 0.4822, 0.4465])
    img_np = np.clip(img_np, 0, 1)
    ax.imshow(img_np)
    ax.set_title(classes[label])
    ax.axis("off")
plt.suptitle("CIFAR-10 Samples")
plt.tight_layout()
plt.savefig("../output/9_4_3_cifar_samples.png", dpi=100, bbox_inches="tight")
plt.show()
```

```python
# Define CNN
class SimpleCNN(nn.Module):
    def __init__(self):
        super().__init__()
        self.features = nn.Sequential(
            nn.Conv2d(3, 32, 3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),
            nn.Conv2d(32, 64, 3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),
            nn.Conv2d(64, 128, 3, padding=1),
            nn.ReLU(),
            nn.MaxPool2d(2),
        )
        self.classifier = nn.Sequential(
            nn.Flatten(),
            nn.Linear(128 * 4 * 4, 256),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(256, 10),
        )

    def forward(self, x):
        x = self.features(x)
        return self.classifier(x)


cnn = SimpleCNN().to(device)
print(f"CNN Parameters: {sum(p.numel() for p in cnn.parameters()):,}")
```

```python
# Train CNN (5 epochs)
cnn_criterion = nn.CrossEntropyLoss()
cnn_optimizer = optim.Adam(cnn.parameters(), lr=0.001)

cnn_losses = []
for epoch in range(5):
    cnn.train()
    running_loss = 0
    for data, target in cifar_train_loader:
        data, target = data.to(device), target.to(device)
        cnn_optimizer.zero_grad()
        loss = cnn_criterion(cnn(data), target)
        loss.backward()
        cnn_optimizer.step()
        running_loss += loss.item()

    avg = running_loss / len(cifar_train_loader)
    cnn_losses.append(avg)
    print(f"Epoch {epoch + 1}/5 | Loss: {avg:.4f}")
```

```python
# Evaluate CNN
cnn.eval()
correct, total = 0, 0
with torch.no_grad():
    for data, target in cifar_test_loader:
        data, target = data.to(device), target.to(device)
        pred = cnn(data).argmax(dim=1)
        correct += (pred == target).sum().item()
        total += target.size(0)

cnn_acc = correct / total
print(f"\nCIFAR-10 Test Accuracy: {cnn_acc:.4f}")

plt.figure(figsize=(7, 4))
plt.plot(range(1, 6), cnn_losses, marker="o")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title(f"CIFAR-10 CNN Training (Acc: {cnn_acc:.2%})")
plt.grid(True, alpha=0.3)
plt.savefig("../output/9_4_4_cifar_loss.png", dpi=100, bbox_inches="tight")
plt.show()
```

## 4. Text Classification with 1D CNN

```python
from torch.utils.data import Dataset
from collections import Counter

# Simple sentiment dataset
texts = [
    ("terrible product would not recommend", 0),
    ("amazing quality love this item", 1),
    ("waste of money fell apart fast", 0),
    ("great value works perfectly", 1),
    ("awful experience never again", 0),
    ("fantastic highly recommend to all", 1),
    ("broke after one day garbage", 0),
    ("excellent build premium quality", 1),
    ("disappointing poor performance", 0),
    ("outstanding product five stars", 1),
    ("useless returned immediately", 0),
    ("wonderful exceeded my expectations", 1),
    ("horrible do not buy this", 0),
    ("perfect exactly what i needed", 1),
    ("defective and cheaply made", 0),
    ("superb quality great design", 1),
]

# Build vocabulary
all_words = " ".join(t for t, _ in texts).split()
vocab = {w: i + 1 for i, (w, _) in enumerate(Counter(all_words).most_common(100))}
vocab_size = len(vocab) + 1
max_len = 8


def encode(text):
    tokens = [vocab.get(w, 0) for w in text.split()][:max_len]
    return tokens + [0] * (max_len - len(tokens))


class SentimentDataset(Dataset):
    def __init__(self, data):
        self.data = [(torch.tensor(encode(t), dtype=torch.long), torch.tensor(l, dtype=torch.long))
                     for t, l in data]

    def __len__(self):
        return len(self.data)

    def __getitem__(self, idx):
        return self.data[idx]


train_data = SentimentDataset(texts[:12])
test_data = SentimentDataset(texts[12:])
```

```python
# 1D CNN for text
class TextCNN(nn.Module):
    def __init__(self, vocab_size, embed_dim=32, num_filters=16):
        super().__init__()
        self.embedding = nn.Embedding(vocab_size, embed_dim, padding_idx=0)
        self.conv3 = nn.Conv1d(embed_dim, num_filters, kernel_size=3, padding=1)
        self.conv4 = nn.Conv1d(embed_dim, num_filters, kernel_size=4, padding=1)
        self.pool = nn.AdaptiveMaxPool1d(1)
        self.fc = nn.Linear(num_filters * 2, 2)
        self.relu = nn.ReLU()

    def forward(self, x):
        x = self.embedding(x).permute(0, 2, 1)  # (batch, embed, seq)
        c3 = self.pool(self.relu(self.conv3(x))).squeeze(-1)
        c4 = self.pool(self.relu(self.conv4(x))).squeeze(-1)
        out = torch.cat([c3, c4], dim=1)
        return self.fc(out)


text_model = TextCNN(vocab_size).to(device)
text_opt = optim.Adam(text_model.parameters(), lr=0.005)
text_loss_fn = nn.CrossEntropyLoss()

# Train
text_model.train()
for epoch in range(50):
    total_loss = 0
    for tokens, label in DataLoader(train_data, batch_size=4, shuffle=True):
        tokens, label = tokens.to(device), label.to(device)
        text_opt.zero_grad()
        loss = text_loss_fn(text_model(tokens), label)
        loss.backward()
        text_opt.step()
        total_loss += loss.item()
    if (epoch + 1) % 10 == 0:
        print(f"Epoch {epoch + 1}/50 | Loss: {total_loss:.4f}")
```

```python
# Evaluate
text_model.eval()
correct = 0
with torch.no_grad():
    for tokens, label in DataLoader(test_data, batch_size=4):
        tokens, label = tokens.to(device), label.to(device)
        pred = text_model(tokens).argmax(dim=1)
        correct += (pred == label).sum().item()

print(f"\nText CNN Accuracy: {correct}/{len(test_data)} = {correct/len(test_data):.2f}")
```
