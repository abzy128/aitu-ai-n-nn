## 2. Control Flow

```python
# 2.1 Deep vs Shallow Copy
import copy

matrix = [[1, 2, 3], [4, 5, 6]]
shallow = copy.copy(matrix)
deep = copy.deepcopy(matrix)

matrix[0][0] = 99
print("Original:", matrix)
print("Shallow copy:", shallow)   # affected
print("Deep copy:", deep)         # unaffected
```

```python
# 2.2 Conditionals
accuracy = 0.87
if accuracy >= 0.95:
    print("Excellent model")
elif accuracy >= 0.85:
    print("Good model")
elif accuracy >= 0.70:
    print("Acceptable model")
else:
    print("Needs improvement")
```

```python
# 2.3 For loop - multiplication table
for i in range(1, 10):
    row = ""
    for j in range(1, i + 1):
        row += f"{j}x{i}={i*j}\t"
    print(row)
```

```python
# 2.4 While loop with break
counter = 0
while counter < 6:
    print(f"Step {counter}")
    counter += 1
    if counter == 4:
        print("Early stop!")
        break
```

## 3. Functions and OOP

```python
# 3.1 Fibonacci sequence
def fibonacci(n):
    """Generate first n Fibonacci numbers."""
    if n <= 0:
        return []
    seq = [0, 1]
    for _ in range(2, n):
        seq.append(seq[-1] + seq[-2])
    return seq[:n]

print("Fibonacci(12):", fibonacci(12))
```

```python
# 3.2 Function with various parameter types
def describe_model(name, framework="PyTorch", *layers, **hyperparams):
    print(f"Model: {name}")
    print(f"Framework: {framework}")
    print(f"Layers: {layers}")
    print(f"Hyperparams: {hyperparams}")

describe_model("CNN", "TensorFlow", 64, 128, 256, lr=0.01, dropout=0.3)
```

```python
# 3.3 Classes and inheritance
class Model:
    def __init__(self, name, params):
        self.name = name
        self.params = params

    def summary(self):
        print(f"{self.name}: {self.params} parameters")

    def train(self):
        print(f"Training {self.name}...")


class ConvNet(Model):
    def __init__(self, name, params, num_layers):
        super().__init__(name, params)
        self.num_layers = num_layers

    def summary(self):
        super().summary()
        print(f"  Layers: {self.num_layers}")


net = ConvNet("ResNet18", 11_700_000, 18)
net.summary()
net.train()
```

```python
# 3.4 Class attributes and methods
class Experiment:
    count = 0

    def __init__(self, name):
        self.name = name
        Experiment.count += 1

    @classmethod
    def total_experiments(cls):
        return cls.count

exp1 = Experiment("baseline")
exp2 = Experiment("augmented")
exp3 = Experiment("regularized")
print(f"Total experiments: {Experiment.total_experiments()}")
```
